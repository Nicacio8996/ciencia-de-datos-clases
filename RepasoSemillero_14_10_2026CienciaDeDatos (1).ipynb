#Iniciando con dataFrame

```python
import pandas as pd

# Datos para el DataFrame se integran en un diccionario
datos = {
    "Nombre": ["Ana", "Bruno", "Carla", "Diego", "Elena"],
    "Edad": [15, 16, 15, 17, 16],
    "Curso": ["3A", "3B", "3A", "4A", "3B"],
    "Horas de estudio": [4, 2, 6, 1, 5],
    "Nota": [8, 6, 9, 5, 8]
}

# Crear el DataFrame
df = pd.DataFrame(datos)

# Mostrar la tabla
df
```

#Cuantas Filas tiene el DataFrame

```python
len(df)
```

#Busca el Valor mas alto de la columna Nota

```python
df['Nota'].max()
```

#Busaca el Vlor Mas Bajo de la  columna nota

```python
df['Nota'].min()
```

#Calcula el promedio de la columna Nota

```python
df["Nota"].mean()
```

#Cuenta cuantas vec3s aparece cada curso

```python
df['Curso'].value_counts()
```

```python
df['Horas de estudio' ].mean()
```
