#Iniciando con dataFrame

```python
import pandas as pd

# Datos para el DataFrame se integran en un diccionario
datos = {
    "Nombre": ["Ana", "Bruno", "Carla", "Diego", "Elena"],
    "Edad": [15, 16, 15, 17, 16],
    "Curso": ["3A", "3B", "3A", "4A", "3B"],
    "Horas de estudio": [4, 2, 6, 1, 5],
    "Nota": [8, 6, 9, 5, 8]
}

# Crear el DataFrame
df = pd.DataFrame(datos)

# Mostrar la tabla
df
```

# Combinar múltiples condiciones (Operadores lógicos)

#Ejemplo con AND (&): Estudiantes del curso "3A" y que hayan estudiado más de 4 horas:

```python
# Nota: Cada condición individual debe ir obligatoriamente entre paréntesis ()
estudiantes_3a_estudiosos = df[(df['Curso'] == '3A') & (df['Horas de estudio'] > 4)]

estudiantes_3a_estudiosos
```

#Ejemplo con OR (|): Estudiantes que tengan una nota de 9 o cuya edad sea de 17 años:

```python
estudiantes_especiales = df[(df['Nota'] == 9) | (df['Edad'] == 17)]

estudiantes_especiales
```

#Promedio de notas únicamente de los alumnos del curso 3A
